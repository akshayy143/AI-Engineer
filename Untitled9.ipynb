```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

```

```python
(x_train, y_train) , (x_test, y_test) = tf.keras.datasets.mnist.load_data()
```

#creating model

```python
x_train  28*28 pixel
y_train 0 to 9 digit
```

```python
model = models.Sequential([
layers.Flatten(input_shape=(28,28)), #input layer
layers.Dense(128, activation='relu'), #hidden layer
layers.Dense(10, activation='softmax') #output layer
])
```

Dense---> no of neuron

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])   #The optimizer is the algorithm that updates the neural network's weights and biases to reduce errors
```

```python
history = model.fit(x_train, y_train, epochs=5)
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print(f'Test accuracy: {test_accuracy * 100:.2f}%')
```

```python
predictions = model.predict(x_test)
```

```python
history = model.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))

# 2. Extract and plot the loss data
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss') # Optional: removes if no validation data

# 3. Add your labels, legend, and title
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Loss vs Epoch')
plt.legend() # Shows which line is training vs validation

# 4. Display the finished plot
plt.show()
```
